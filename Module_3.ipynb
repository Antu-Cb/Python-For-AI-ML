## String, Indexing & Slicing

```python
text = "Hello World"
print(text, type(text))

prompt = """What is your name?
How are you?"""

print(prompt)

## Indexing

print(text[0])
print(text[-1]) # backward indexing

## Slicing

text_sliced = text[0:5]
print(text_sliced)

text_sliced_1 = text[0:5:2] # [start:end:step]
print(text_sliced_1)
```

## String Methods

```python
str = "Hello World"

print(len(str))
print("Hello" in str)
print(str.upper())
print(str.lower())
print(str.capitalize())
print(str.find("World"))
print(str.count("World"))
print(str.replace("World", "Python"))
```

## String Splitting, Joining, Formatting

```python
## Splitting

prompt = "What is Python?"
tokens = prompt.split()
print(tokens)

## Joining

sentence = " ".join(tokens)
print(sentence)

## Formatting

name = input("Enter name: ")
print(f"Hello, {name}")
print(f"Hello, {name.upper()}")
```

## List and List Methods

```python
numbers = [100, 20, 30, 40, 50]
print(numbers, type(numbers))

## Access

print(numbers[2])
numbers[2] = 60
print(numbers[2]) # list is mutable

## Slicing

new_list = numbers[0:3]
print(new_list)

## Modification

numbers.append(70) # value will add at the end
print(numbers)

numbers.insert(0, 10) # value will add at the specific index
print(numbers)

numbers.pop() # value will remove from the end
print(numbers)

numbers.remove(100) # specific value will be removed
print(numbers)
```

## List in a List/ 2D List

```python
list_2D = [[10, 20, 30], ["a", "b", "c"], [1, 2, 3]]

print(list_2D[0])
print(list_2D[0][1])
```

## List as Stack and Queue

```python
nums = [100, 20, 30, 10]
nums.sort()
print(nums)

nums.reverse() # must need to do sort before applying reverse method
print(nums)

## List as Stack

stack = []

stack.append(1)
stack.append(2)
stack.append(3)
stack.append(4)

print(stack)

stack.pop()
print(stack)

## List as Queue

queue = []

queue.append(10)
queue.append(20)
queue.append(30)
queue.append(40)

print(queue)

queue.pop(0) # Queue te first value pop hobe
print(queue)

print(queue[0]) # new top value
```

## List Comprehension

```python
even = []

for i in range(1,100):
  if(i%2 == 0):
    even.append(i)

print(even)

## list comprehension

even = [i for i in range(1, 100) if(i%2 == 0)]
print(even)

## List comprehension with a list of string

fruits = ['apple', 'orange', 'lichi']

upper_fruits = [fruit.upper() for fruit in fruits]
print(upper_fruits)
```
