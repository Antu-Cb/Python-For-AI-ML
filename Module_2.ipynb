## Conditional Statements

```python
raining = True

if(raining == True):
  print("Weather: Rainy")
else:
  print("Weather: Sunny")

## Nested if-else

muddy = False

if(raining == True):
  if(muddy == True):
    print("Rainy & Muddy")
  else:
    print("Only Rainy")

else:
  print("Sunny")

## elif Statement

mark = int(input("Enter mark: "))

if(mark>=80):
  print("A+")
elif(mark>=70):
  print("A")
else:
  print("A-")
```

# For loop in python

```python
for i in range(10):
  print(i)

print("\n")

# range(start, end, step)
for i in range(1, 11, 2):
  print(i)

print("\n")

# Nested for loop

for i in range(6):
  for j in range(6):
    print(i, j)
```

# While loop

```python
count = 1
while(count<11):
  print(count, "Hello world")
  count += 1
```

## Break & Continue Statement

```python
for i in range(1,11):
  if(i%2 == 0):
    continue # even num pele skip
  print(i)

## Break statement

accuracy = 30

for i in range(100):
  accuracy += 1

  if(accuracy == 100):
    break
  print(accuracy)
```

## Problem Solve

```python
## Min-Max problem

nums = input().split()
# List Comprehension
nums_int = [int(num) for num in nums] # converts into int

max = nums_int[0]
min = nums_int[0]

for num in nums_int:
  if(num > max):
    max = num
  if(num < min):
    min = num

print(min, max)
```

```python
## Digits

T = int(input())
nums = []
# Taking input
for i in range(T):
  num = int(input())
  nums.append(num)

# Calculation
for num in nums:
  if(num == 0):
    print(num)
    continue

  while(num!=0):
    print(num%10, end=" ")
    num = num//10
  print()


```
